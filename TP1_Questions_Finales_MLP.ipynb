# Questions Finales — MLP et modèles classiques

Ce fichier répond aux questions de synthèse sur les performances du MLP par rapport aux modèles classiques et aux CNN.

**1. Dans quels cas le MLP a-t-il mieux performé que les modèles classiques ?**

Le MLP surpasse les modèles classiques (comme la régression logistique, SVM ou k-NN) lorsque les données présentent des relations non linéaires complexes ou de haut niveau, difficiles à capturer par des modèles linéaires ou à base de distance. Sur MNIST, le MLP atteint une très bonne accuracy (>97%), supérieure à la plupart des modèles classiques sans ingénierie de features poussée. Il excelle aussi lorsque le volume de données est suffisant pour entraîner ses nombreux paramètres.

**2. Dans quels cas les modèles classiques restent-ils meilleurs ?**

Les modèles classiques restent meilleurs lorsque les données sont peu nombreuses, très bruitées, ou que la structure du problème est simple (linéaire ou faiblement non linéaire). Ils sont aussi préférés pour leur rapidité d'entraînement, leur faible besoin en ressources, et leur interprétabilité. Sur de petits jeux de données ou des problèmes tabulaires simples, une régression logistique ou un SVM bien réglé peut surpasser un MLP qui risque d'overfitter.

**3. Pourquoi le MLP est-il moins performant que les CNN sur MNIST ?**

Le MLP traite chaque pixel indépendamment, sans exploiter la structure spatiale de l'image. Il ne reconnaît pas les motifs locaux (bords, textures) et ne bénéficie pas de l'invariance locale. Les CNN, grâce aux convolutions et au partage de poids, capturent efficacement les motifs visuels et la hiérarchie spatiale, ce qui leur permet d'atteindre une bien meilleure performance sur des images comme MNIST.

**4. Quel serait le principal avantage d’un CNN ici ?**

Le principal avantage d'un CNN sur MNIST est sa capacité à exploiter la structure spatiale des images. Il apprend automatiquement des filtres adaptés aux motifs locaux (traits, courbes) et généralise mieux aux variations d'écriture. Cela se traduit par une meilleure accuracy, une robustesse accrue et une efficacité paramétrique supérieure par rapport à un MLP.
